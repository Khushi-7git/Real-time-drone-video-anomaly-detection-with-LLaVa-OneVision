```python

"""
AHC Visual Intelligence Hackathon — Fast zero-shot pipeline
LLaVA-OneVision-0.5B for real-time-style drone anomaly classification.

Strategy: NO fine-tuning. Zero-shot multi-class prompting on sampled frames
+ a lightweight temporal rule layer. Designed to run end-to-end on a Colab
T4 in well under an hour so you have time left for tuning prompts/thresholds.

Expected data layout (from your screenshots):
  train/<class_name>/videos/*.mp4
  train/<class_name>/videos.csv
  train/<class_name>/ground_truth.csv
  test/<class_name>/videos/*.mp4
  test/<class_name>/videos.csv
  test/<class_name>/ground_truth.csv
"""

import os
import re
import glob
import json
import pandas as pd
import numpy as np
import torch
from decord import VideoReader, cpu
from transformers import AutoProcessor, LlavaOnevisionForConditionalGeneration, BitsAndBytesConfig

# ---------------------------------------------------------------------------
# 0. CONFIG
# ---------------------------------------------------------------------------
DATA_ROOT = "/content/drive/MyDrive/dataset"   # <-- change to your path
SPLIT = "train"                                 # "train" or "test"
NUM_FRAMES = 2                                # fewer frames = faster, less accurate
MODEL_ID = "llava-hf/llava-onevision-qwen2-0.5b-ov-hf"  # small = fast; swap to -7b if you have time/VRAM
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

CLASSES = [
    "traffic_congestion",
    "stalled_or_broken_down_vehicle",
    "vehicle_blocking_traffic",
    "traffic_accident",
    "smoke",
    "fire",
    "waterlogging_or_flood",
    "loitering_or_suspicious_presence",
    "road_spill_or_debris",
    "fighting_or_violence",
    "wrong_way_driving",
    "normal",
]

# Classes where a single-frame cue is enough to trigger (fast events)
INSTANT_CLASSES = {"traffic_accident", "fire", "smoke", "fighting_or_violence"}
# Classes that need to be seen across MULTIPLE sampled frames to count
# (slow-building / persistent conditions)
PERSISTENT_CLASSES = {
    "traffic_congestion", "stalled_or_broken_down_vehicle",
    "waterlogging_or_flood", "loitering_or_suspicious_presence",
    "road_spill_or_debris",
}
PERSISTENCE_FRAC = 0.5  # class must appear in >=50% of frame-level votes

# ---------------------------------------------------------------------------
# 1. LOAD MODEL (4-bit for speed/VRAM headroom)
# ---------------------------------------------------------------------------
def load_model():
    bnb_config = BitsAndBytesConfig(load_in_4bit=True, bnb_4bit_compute_dtype=torch.float16)
    processor = AutoProcessor.from_pretrained(MODEL_ID)
    model = LlavaOnevisionForConditionalGeneration.from_pretrained(
        MODEL_ID,
        quantization_config=bnb_config,
        torch_dtype=torch.float16,
        device_map="auto",
    )
    model.eval()
    return model, processor


# ---------------------------------------------------------------------------
# 2. FRAME SAMPLING
# ---------------------------------------------------------------------------
def sample_frames(video_path, num_frames=NUM_FRAMES):
    vr = VideoReader(video_path, ctx=cpu(0))
    total = len(vr)
    if total == 0:
        return []
    idx = np.linspace(0, total - 1, num_frames).astype(int)
    frames = vr.get_batch(idx).asnumpy()  # (num_frames, H, W, 3)
    return [frames[i] for i in range(frames.shape[0])]


# ---------------------------------------------------------------------------
# 3. PROMPT + SINGLE-FRAME CLASSIFICATION
# ---------------------------------------------------------------------------
PROMPT_TEMPLATE = (
    "You are monitoring a live drone feed over a city for public-safety anomalies. "
    "Look at this frame and classify it into EXACTLY ONE of these categories:\n"
    + ", ".join(CLASSES) + ".\n"
    "If nothing unusual is happening, answer 'normal'. "
    "Respond with only the category name, nothing else."
)

def classify_frame(model, processor, frame_np):
    conversation = [
        {
            "role": "user",
            "content": [
                {"type": "image"},
                {"type": "text", "text": PROMPT_TEMPLATE},
            ],
        }
    ]
    prompt = processor.apply_chat_template(conversation, add_generation_prompt=True)
    inputs = processor(images=frame_np, text=prompt, return_tensors="pt").to(DEVICE, torch.float16)

    with torch.no_grad():
        out = model.generate(**inputs, max_new_tokens=10, do_sample=False)

    text = processor.decode(out[0], skip_special_tokens=True).lower()
    return map_to_class(text)


def map_to_class(text):
    text = text.strip().lower()
    for c in CLASSES:
        if c.replace("_", " ") in text or c in text:
            return c
    # fuzzy fallback: word-overlap match
    best, best_score = "normal", 0
    words = set(re.findall(r"[a-z]+", text))
    for c in CLASSES:
        c_words = set(c.split("_"))
        score = len(words & c_words)
        if score > best_score:
            best, best_score = c, score
    return best


# ---------------------------------------------------------------------------
# 4. VIDEO-LEVEL DECISION (temporal rule layer, no extra model calls)
# ---------------------------------------------------------------------------
def classify_video(model, processor, video_path):
    frames = sample_frames(video_path)
    if not frames:
        return "normal"

    votes = [classify_frame(model, processor, f) for f in frames]
    counts = pd.Series(votes).value_counts()

    # Instant classes: any single hit wins immediately
    for c in INSTANT_CLASSES:
        if counts.get(c, 0) >= 1:
            return c

    # Persistent classes: need majority presence across frames
    n = len(votes)
    for c in PERSISTENT_CLASSES:
        if counts.get(c, 0) / n >= PERSISTENCE_FRAC:
            return c

    # Otherwise fall back to the mode vote (covers remaining classes / normal)
    return counts.idxmax()


# ---------------------------------------------------------------------------
# 5. RUN OVER DATASET FOLDER STRUCTURE
# ---------------------------------------------------------------------------
def run(split=SPLIT):
    model, processor = load_model()
    class_dirs = [d for d in glob.glob(os.path.join(DATA_ROOT, split, "*")) if os.path.isdir(d)]

    results = []
    for class_dir in class_dirs:
        true_class_name = os.path.basename(class_dir)
        gt_path = os.path.join(class_dir, "ground_truth.csv")
        gt_df = pd.read_csv(gt_path) if os.path.exists(gt_path) else None

        video_files = glob.glob(os.path.join(class_dir, "videos", "*.mp4"))
        video_files = video_files[:5]
        for vf in video_files:
            pred = classify_video(model, processor, vf)
            results.append({
                "video": os.path.basename(vf),
                "folder_class": true_class_name,
                "predicted_class": pred,
            })
            print(f"[{true_class_name}] {os.path.basename(vf)} -> {pred}")

    out_df = pd.DataFrame(results)
    out_df.to_csv(f"{split}_predictions.csv", index=False)
    return out_df


# ---------------------------------------------------------------------------
# 6. QUICK METRICS (folder name as proxy ground truth if ground_truth.csv
#    schema is per-frame/per-timestamp rather than per-video — adjust the
#    join once you inspect the actual columns in ground_truth.csv)
# ---------------------------------------------------------------------------
def evaluate(pred_df):
    correct = (pred_df["predicted_class"] == pred_df["folder_class"]).sum()
    total = len(pred_df)
    acc = correct / total if total else 0
    print(f"Video-level accuracy vs folder label: {acc:.3f} ({correct}/{total})")
    print(pred_df.groupby("folder_class").apply(
        lambda g: (g["predicted_class"] == g["folder_class"]).mean()
    ))


if __name__ == "__main__":
    preds = run(SPLIT)
    evaluate(preds)
```
